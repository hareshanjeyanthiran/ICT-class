```python
import matplotlib.pyplot as plt
x = [0,5,10,15,20,25,30]
y1=[10,15,20,20,30,15,0]
y2=[10,12,15,12,20,10,0]

plt.plot(x,y1,linestyle='dashed',marker='D',color='blue',label='Velocity-1')
plt.plot(x,y2,linestyle='dotted',marker='*',color='green',label='Velocity-2')
plt.title('Velocity-Time Graph')
plt.xlabel('Velocity(m/s)')
plt.ylabel('Time(s)')
plt.xlim(5,20)
plt.ylim(5,30)
plt.legend()
plt.show()
```

```python
import numpy as np

x=np.arange(1,11,1)
print(x)

y1=(2*x)+1
y2=(2*x**x)+2

plt.plot(x,y1,'g',linewidth=3,label='y=2x+1')
plt.plot(x,y2,'r',linewidth=3,label='y=2x^x+2')
plt.legend()
plt.show()
```

```python
x=[1,2,3,4]
y1 = [2,4,6,8]
y2 = [3,6,9,12]
plt.plot(x,y1,'g',linewidth=3,label='p1')
plt.plot(x,y2,'r',linewidth=3,label='p2')

plt.fill_between(x,y1,y2,color='orange',alpha=0.5)
plt.legend()
plt.show()
```
